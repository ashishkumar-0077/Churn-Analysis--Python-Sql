```python
# Import Libraries
import numpy as np
import pandas as pd
```

```python
# Load Datase
df = pd.read_excel("Churn_Unclean_Project.xlsx")
```

```python
# first Look
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.shape
```

```python
# Check Missing Values
df.isnull().sum()
```

```python
# Check Duplicated Records
df.duplicated().sum()
```

```python
# Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
# Replace Dirty Values
df.replace(['N/A','NULL','',' '],np.nan,inplace=True)
```

```python
# Remove Extra space
df = df.apply(lambda x:x.str.strip() if x.dtypes =='object' else x)
```

```python
df.info()
```

```python
# Proper Cae
df['Customer_Name'] = df['Customer_Name'].str.title()
df['State'] = df['State'].str.title()
df['City'] = df['City'].str.title()
df['Subscription_Type'] = df['Subscription_Type'].str.title()
df['Contract_Type'] = df['Contract_Type'].str.title()
df['Churn'] = df['Churn'].str.title()
df['Gender'] = df['Gender'].str.title()
```

```python
df['Subscription_Type'].value_counts()
```

```python
# Standerdize churn column
df['Churn'].value_counts()
```

```python
# Convert Numeric Column
cols = ['Age','Tenure_Months','Monthly_Charges','Total_Charges']

for c in cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")
```

```python
df.describe()
```

```python
# Remove Invaild Age
df = df[(df['Age']>=18)&(df['Age']<=100)]
```

```python
# Remove Negative Charges
df = df[df['Monthly_Charges']>=0]
df = df[df['Total_Charges']>=0]
```

```python
df.shape
```

```python
# Convert Date
df['Last_Interaction_Date'] = pd.to_datetime(df['Last_Interaction_Date'], errors='coerce')
```

```python
df.info()
```

```python
# Fill Blank Values
df['Tech_Support'] = df['Tech_Support'].fillna('No')
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')
df['Monthly_Charges'] = df['Monthly_Charges'].fillna(df['Monthly_Charges'].mean())
df['Tenure_Months'] = df['Tenure_Months'].fillna(df['Tenure_Months'].mean())
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Last_Interaction_Date'] = df['Last_Interaction_Date'].fillna(df['Last_Interaction_Date'].median())
```

```python
# Featuren Engeenering
# Customer Lifetime Value
df['Customer_Values'] = df['Monthly_Charges']*df['Tenure_Months']
```

```python
df.head()
```

```python
# Monthly Revenue
df['Monthly_Revenue'] = df['Monthly_Charges']
```

```python
# Tenure Group["0-12","13-24","25-48","49-72"]
bins = [0,12,24,48,72]
label=["0-12 month", "13-24 month", "25-48 month", "49-72 month"]

df['Tenure_Group'] = pd.cut(df['Tenure_Months'],bins=bins, labels=label)
```

```python
df.head()
```

```python
# Senior Flag
df['Senior Flag'] = np.where(df['Age']>=60, "Senior","Adult")
```

```python
df.head()
```

```python
# Churn Falg
df['Churn_Flag'] = df['Churn'].map({"Yes":1,"No":0})
```

```python
df.head()
```

```python
# Final Check
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# Export Clean csv file
df.to_csv("Clean_Churn_Data.csv", index=False)
```
